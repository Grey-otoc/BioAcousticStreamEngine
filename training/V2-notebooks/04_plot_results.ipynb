# 04 — Plot Results
Plot training and validation mAP metrics over epochs to highlight any underfitting or overfitting issues.

**Kernel:** `Python (orthoptera-training)`  
**Prerequisites:** Run `02_prepare_labels.ipynb` first to generate train, test, and val CSVs, train the model, and manually record the mAP metrics to plot the results.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# ── Plot Train and Val mAP per epoch ───────────────────────────
# Details severe overfitting, with val mAP stagnant as train mAP reaches near perfection.

data = {
    "Epoch": list(range(30)),
    "Train_mAP": [
        0.834, 0.944, 0.964, 0.973, 0.980, 0.983, 0.985, 0.988, 0.988, 0.990,
        0.993, 0.994, 0.994, 0.995, 0.995, 0.995, 0.995, 0.995, 0.996, 0.995,
        0.997, 0.997, 0.998, 0.998, 0.998, 0.997, 0.997, 0.998, 0.997, 0.998
    ],
    "Val_mAP": [
        0.535, 0.566, 0.541, 0.561, 0.590, 0.566, 0.580, 0.520, 0.546, 0.518,
        0.556, 0.566, 0.566, 0.582, 0.542, 0.546, 0.536, 0.544, 0.549, 0.588,
        0.548, 0.566, 0.572, 0.555, 0.591, 0.555, 0.568, 0.544, 0.572, 0.573
    ]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 6))
plt.grid(True, linestyle="--", alpha=0.6)

plt.plot(df["Epoch"], df["Train_mAP"], label="Training mAP", color="#820319", linewidth=2, marker="s", markersize=4)
plt.plot(df["Epoch"], df["Val_mAP"], label="Validation mAP", color="#033482", linewidth=2, marker="s", markersize=4)

plt.title("mAP Metrics by Epoch (30 Epoch w/ Neg Clips and Resampling)", fontsize=16, pad=15)
plt.xlabel("Epoch", fontsize=12, labelpad=10)
plt.ylabel("Mean Average Precision (mAP)", fontsize=12, labelpad=10)

plt.xticks(df["Epoch"][::2])

plt.legend(loc="center right", fontsize=12, frameon=True, facecolor="white", edgecolor="none")

plt.tight_layout()
plt.show()
```
